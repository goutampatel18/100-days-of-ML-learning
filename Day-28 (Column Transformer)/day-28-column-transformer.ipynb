```python
import numpy as np
import pandas as pd
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('/kaggle/input/datasets/goutampatel18/covid-toy/covid_toy.csv')
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],
                                                test_size=0.2)
```

```python
X_train
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
    ('tnf3',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough')

#in the above code ColumnTransformer is used for the transforming the column and we also pass remainder in which the two options are there 'drop' and 'passthrough' 
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test).shape
```
